```python
import pandas as pd
import requests
from io import StringIO
```

```python
stations = [
    "KNYC", "KCLT", "KCQT", "KHOU", "KIND",
    "KJAX", "KMDW", "KPHL", "KPHX", "KSEA"
]
```

```python
base_url = "https://raw.githubusercontent.com/fivethirtyeight/data/master/us-weather-history/"
dfs = []
for station in stations:
    url = f"{base_url}{station}.csv"
    response = requests.get(url)
    
    df = pd.read_csv(StringIO(response.text))
    df["station"] = station   # add station identifier
    dfs.append(df)
combined_df = pd.concat(dfs, ignore_index=True)
```

```python
combined_df.to_csv("us_weather_history_combined.csv", index=False)
```

```python
with plt.style.context(
    'https://gist.githubusercontent.com/rhiever/d0a7332fe0beebfdc3d5/raw/223d70799b48131d5ce2723cd5784f39d7a3a653/tableau10.mplstyle'
):
    # Filter date range
    weather_data_subset = weather_data[
        (weather_data["date"] >= datetime(2014, 7, 1)) &
        (weather_data["date"] < datetime(2015, 7, 1))
    ].copy()
    weather_data_subset["day_order"] = range(len(weather_data_subset))
    day_order = weather_data_subset["day_order"]
    record_max_temps = weather_data_subset["record_max_temp"]
    record_min_temps = weather_data_subset["record_min_temp"]
    average_max_temps = weather_data_subset["average_max_temp"]
    average_min_temps = weather_data_subset["average_min_temp"]
    actual_max_temps = weather_data_subset["actual_max_temp"]
    actual_min_temps = weather_data_subset["actual_min_temp"]
    fig, ax1 = plt.subplots(figsize=(15, 7))
    # Record highs/lows
    ax1.bar(
        day_order,
        record_max_temps - record_min_temps,
        bottom=record_min_temps,
        color="#C3BBA4",
        width=1,
        edgecolor="none"
    )
    # Average highs/lows
    ax1.bar(
        day_order,
        average_max_temps - average_min_temps,
        bottom=average_min_temps,
        color="#9A9180",
        width=1,
        edgecolor="none"
    )
    # Actual highs/lows
    ax1.bar(
        day_order,
        actual_max_temps - actual_min_temps,
        bottom=actual_min_temps,
        color="#5A3B49",
        width=1,
        edgecolor="black",
        linewidth=0.5
    )
    # New records
    new_max_records = weather_data_subset[
        weather_data_subset["actual_max_temp"] >= weather_data_subset["record_max_temp"]
    ]

    new_min_records = weather_data_subset[
        weather_data_subset["actual_min_temp"] <= weather_data_subset["record_min_temp"]
    ]
    ax1.scatter(
        new_max_records["day_order"] + 0.5,
        new_max_records["actual_max_temp"] + 1.25,
        s=15,
        color="#d62728",
        zorder=10
    )
    ax1.scatter(
        new_min_records["day_order"] + 0.5,
        new_min_records["actual_min_temp"] - 1.25,
        s=15,
        color="#1f77b4",
        zorder=10
    )
    # Axis formatting
    ax1.set_ylim(-15, 111)
    ax1.set_xlim(-5, 370)
    ax1.set_yticks(range(-10, 111, 10))
    ax1.set_yticklabels([f"{x}°" for x in range(-10, 111, 10)])
    ax1.set_ylabel("Temperature (°F)", fontsize=12)
    # Month labels
    month_starts = weather_data_subset[weather_data_subset["date"].dt.day == 1]
    month_indices = list(month_starts["day_order"])
    month_names = list(month_starts["date"].dt.strftime("%B"))
    month_names[0] += "\n'14"
    month_names[6] += "\n'15"
    month_indices.append(weather_data_subset["day_order"].iloc[-1])
    month_names.append("July")
    ax1.set_xticks(month_indices)
    ax1.set_xticklabels(month_names)
    ax2 = ax1.twiny()
    ax2.set_xlim(ax1.get_xlim())
    ax2.set_xticks(month_indices)
    ax2.set_xticklabels(month_names)
    ax1.grid(False)
    plt.tight_layout()
    plt.show()
```
